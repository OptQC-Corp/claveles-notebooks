# Two-Mode EPR State: Generation and Quantum Correlations

## Introduction

This demonstration illustrates the properties of a two-mode **EPR (Einstein–Podolsky–Rosen) state**—one of the most fundamental resources in continuous-variable (CV) quantum information processing. We will explore this by constructing a bosonic quantum circuit to generate the state and observe the quantum correlations in quadratures between the two modes, witnessing the existence of nonclassical entanglement that clearly demonstrates the quantum nature of the state.

## Background

### What Is an EPR State?

At its core, an EPR state represents the most profound and counterintuitive phenomenon in quantum mechanics: quantum entanglement. Imagine two parts of a physical system that are deeply connected, acting as a single, unified entity regardless of the physical distance separating them. In an EPR state, these subsystems are so perfectly intertwined that you cannot describe the state of one without simultaneously describing the other.

The name "EPR" comes from three eminent physicists—Albert Einstein, Boris Podolsky, and Nathan Rosen—who introduced the concept of **entanglement** in a famous 1935 paper [\[1\]](#References). Ironically, they did not propose this state to celebrate the wonders of quantum mechanics. Instead, they used it as a thought experiment to argue that quantum theory inherently predicts a counter-intuitive phenomenon and thus must be fundamentally incomplete. Einstein famously dismissed this phenomenon as "spooky action at a distance," refusing to believe that such a bizarre connection could exist in nature.

To understand why Einstein was so troubled, it helps to compare the EPR state with everyday classical physics. Imagine you separate a pair of shoes, putting the left shoe in one box and the right in another, and send one of the boxes to a friend living at the opposite end of the world. Opening your box immediately tells you what is in the other, even without confirming what your friend received. However, in this scenario, the shoes' identities were fixed from the very beginning, long before anyone opened the boxes. An EPR pair is fundamentally different. When you distribute an EPR pair to two distinct parties, neither of the quantum systems has a definite state before you observe it; they exist in a suspended state of probability called **superposition**. Yet, the exact moment you measure one half of the pair, its state becomes definite, and the other half's state instantly snaps into a corresponding certainty—no matter how many light-years away it might be.

For decades, the EPR state remained a subject of intense philosophical debate. Even Einstein rejected the idea that an "indefinite" quantum state could exist in the real world of physics. He tried to understand the stochastic behavior of the quantum world as a predetermined probability distribution, implying that quantum mechanics was merely an incomplete description of reality rather than a fundamentally random phenomenon. Today, however, our understanding has undergone a dramatic paradigm shift. The search for local realism eventually led to John Bell's formulation of Bell's inequalities in 1964 [\[2\]](#References) and their subsequent experimental demonstrations [\[3,4\]](#References), which provided a way to experimentally prove the non-classical nature of quantum physics. What was once considered a mere paradox is now the fundamental "resource" powering the modern quantum technology revolution. These so-called **entangled states** are the essential building blocks for unbreakable quantum cryptography, quantum teleportation, and the unprecedented processing power of the quantum computers we exploit today.

### Definitions

The ideal EPR state is defined as a two-mode bosonic state that satisfies the following equalities:

$$
\hat{x}_0 - \hat{x}_1 = 0, \quad \hat{p}_0 + \hat{p}_1 = 0,
$$

where the subscripts 0 and 1 denote the two CV modes. These equations indicate that the modes share perfectly correlated $x$-quadratures and perfectly anti-correlated $p$-quadratures. In other words, if one measures $x_0$ of mode 0 in an EPR state and obtains an outcome $a$, then measuring $x_1$ in mode 1 must yield exactly the same value $a$, exhibiting a clear correlation in the position basis $x$. The same applies to the momentum quadrature $p$, but with the sign flipped; i.e., when observing $a$ for $p_0$, one must find $-a$ for $p_1$. The EPR state can also be described using quadrature expansions as follows:

$$
\ket{\mathrm{EPR}} = \int_{-\infty}^{\infty} da \, \ket{x_0 = a} \otimes \ket{x_1 = a} = \int_{-\infty}^{\infty} da \, \ket{p_0 = a} \otimes \ket{p_1 = -a}.
$$

Using the Dirac delta function $\delta$, these expansions can be understood as a state with a wave function $\psi(x_0, x_1) = \delta(x_0 - x_1)$ in the position basis, whose Fourier transform gives the momentum expression $\tilde{\psi}(p_0, p_1) = \delta(p_0 + p_1)$. As seen from the singular nature of these wave functions, the ideal EPR state requires infinite energy, making it physically unrealizable and impossible to formally define as a physical state. It is rather considered only as an ideal limit of physical approximations. (Recall that the same argument applies to unphysical quadrature eigenstates versus physical squeezed states; see the [CV addition notebook](./add.ipynb).) Because physical states must have finite energy, in practice we use a finitely correlated physical EPR state $\ket{\mathrm{EPR}_r}$ with squeezing parameter $r$, defined as:

$$
\ket{\text{EPR}_r} = \frac{1}{\sqrt{\pi \hbar}} \iint_{-\infty}^{\infty} dx_0 dx_1 \exp\left[ - \frac{e^{-2r}}{4\hbar} (x_0 + x_1)^2 - \frac{e^{2r}}{4\hbar} (x_0 - x_1)^2 \right] \ket{x_0}\otimes\ket{x_1} 
$$

which reproduces the ideal EPR state in the limit of $r\to\infty$. The strength of quantum entanglement in $\ket{\text{EPR}_r}$ can be quantified by the variances of the quadrature combinations:

$$
\Delta (x_0-x_1) = \Delta (p_0+p_1) = \sqrt{\hbar}e^{-r}.
$$

In a single qumode, the position and momentum operators satisfy the canonical commutation relation $[\hat{x},\hat{p}]=i\hbar$, and thus the uncertainty relation $\Delta x \cdot \Delta p \geq \hbar/2$ follows. However, the linear combinations of the quadrature operators of the two qumodes commute, i.e., $[\hat{x}_0-\hat{x}_1, \hat{p}_0+\hat{p}_1]=0$, enabling both variances to be arbitrarily suppressed in theory. Also note that at the other limit, $r=0$, the EPR state reduces to a simple product of two vacuum states $\ket{0}_0\otimes\ket{0}_1$ with no entanglement. Thus, the squeezing parameter $r$ smoothly interpolates between an unentangled trivial state ($r=0$) and the maximally correlated ideal EPR state ($r\to\infty$).

### Generation by Beamsplitter

The physical EPR state $\ket{\text{EPR}_r}$ can be generated by interacting two single-mode squeezed vacuum states (both with squeezing level $r$) via a 50:50 beamsplitter gate—with mode 0 being $x$-squeezed and mode 1 being $p$-squeezed. In the standard convention, the beamsplitter unitary operation is defined as:

$$
\hat{B}(\theta, \phi) = \exp\!\left[\theta\!\left(e^{i\phi}\hat{a}_0\hat{a}_1^\dagger - e^{-i\phi}\hat{a}_0^\dagger\hat{a}_1\right)\right],
$$

where $\theta = \pi/4$ yields a 50:50 splitting ratio and $\phi$ controls the relative phase. Choosing $\theta = \pi/4$ and $\phi = \pi$ directly implements the EPR coupling without requiring additional phase rotations. At this specific phase, the off-diagonal coupling factor $e^{\pm i\phi}$ in the beamsplitter unitary becomes $-1$, ensuring that $\hat{x}_0$ couples exclusively to $\hat{x}_1$ and $\hat{p}_0$ couples exclusively to $\hat{p}_1$ (without mixing conjugate quadratures). The quadrature operators transform in the Heisenberg picture as follows:

$$
(\hat{x}_0, \hat{p}_0, \hat{x}_1, \hat{p}_1)
\xrightarrow{\hat{B}(\pi/4,\,\pi)}
(\hat{x}_0', \hat{p}_0', \hat{x}_1', \hat{p}_1')
=\tfrac{1}{\sqrt{2}}\big(\hat{x}_0 + \hat{x}_1,\; \hat{p}_0 + \hat{p}_1,\; \hat{x}_1 - \hat{x}_0,\; \hat{p}_1 - \hat{p}_0\big),
$$

where the primed operators denote the output quadratures after the beamsplitter. When mode 0 is initially $x$-squeezed and mode 1 is $p$-squeezed, the quadratures of these squeezed states are represented as the vacuum quadrature operators rescaled by a factor of $e^{-r}$, as

$$
 \hat{x}_0 = e^{-r} \hat{x}_{0,\mathrm{vac}}, \;\hat{p}_1 = e^{-r} \hat{p}_{1,\mathrm{vac}}.
$$

Thus, the variances of the difference and sum quadratures vanish:

$$
\hat{x}_0' - \hat{x}_1' = \sqrt{2}\,e^{-r} \hat{x}_{0,\mathrm{vac}} \to 0, \qquad \hat{p}_0' + \hat{p}_1' = \sqrt{2}\,e^{-r} \hat{p}_{1,\mathrm{vac}} \to 0
$$

in the limit of $r\to\infty$.

### Marginal Statistics of a Single Mode

A fundamental feature of quantum entanglement is that the complete state of a composite system cannot be divided into independent states of its individual parts. Consequently, if we restrict our observation to just one subsystem and completely ignore—or mathematically "trace out"—the rest of the system, we inherently lose crucial quantum information contained in the entangled state. Without the context of its entangled counterpart, the subsystem we observe can no longer be described by a pure quantum state; instead, it manifests as a "mixed state." A mixed state is fundamentally characterized by a statistical mixture of pure states, appearing simply as a collection of classical probabilities instead of a quantum superposition.

For an EPR state, tracing out one mode leaves the remaining mode in a so-called **thermal state**, expanded in the Fock basis as follows:

$$
\rho_i = \frac{1}{\cosh^2 r} \sum_{n=0}^{\infty} (\tanh r)^{2n} \ket{n}_i\bra{n}_i
$$

where the subscript $i$ denotes the mode that is not traced out; see [Fock-Number Measurement on an EPR State](./epr3.ipynb) for the deriviation. (Note that, as can be seen from the position expansion, an EPR state is symmetric under mode permutation, and thus tracing out either of the modes outputs the same thermal state.) The reduced density matrix becomes a Gibbs state $\exp(-\beta \hat{n})/\mathrm{Tr}(\exp(-\beta \hat{n}))$ with $\beta = 2\log(\coth r)$, where $\hat{n}$ denotes the number operator that designates the Fock number. This state realizes the Bose–Einstein distribution that appears in classical statistical mechanics. This showcases a profound connection between quantum entanglement and thermodynamics, where the loss of information from tracing out one half of a pure, bipartite entangled state manifests locally as classical thermal noise. In phase space, the single-mode marginal Wigner function presents as a broad, isotropic Gaussian with equal uncertainty $\Delta x =\Delta p = \sqrt{\frac{\hbar \cosh(2r)}{2}}$ in both quadratures—a definitive hallmark of a thermal state that exhibits neither purity nor squeezing.

### Conditional State After Measurement

According to the projection postulate, measuring one mode of an EPR pair via homodyne detection projects the remaining mode into a squeezed state that is displaced by an amount proportional to the measurement outcome. This is a direct consequence of the strong $\hat{x}_0 = \hat{x}_1$ correlation: acquiring knowledge about $x_0$ immediately determines the state configuration of $x_1$. With a physical EPR state, however, the entanglement is not perfect, and thus the homodyne projection does not project the residual mode into a precise position eigenstate. Specifically, after obtaining an outcome $a$ in the homodyne measurement for mode 0, the remaining state in mode 1 is represented as

$$
\int_{-\infty}^{\infty} dx_1 \left( \frac{\cosh(2r)}{\pi \hbar} \right)^{\frac{1}{4}} \exp\left[ - \frac{\cosh(2r)}{2\hbar} \left( x_1 - a \tanh(2r) \right)^2 \right]\ket{x_1},
$$

which is an $x$-squeezed state with $\Delta x_1 = \sqrt{\frac{\hbar}{2\cosh(2r)}}$ displaced by an amount $a \tanh(2r)$. As expected, the stronger the quantum correlation is ($r\to\infty$), the more tightly the resulting state is localized around $x_1\approx a$ with a smaller variance.

## 1. Setup

First, let us import the required libraries and define the common simulation parameters.

```python
from math import pi

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from claveles.circuit import CircuitRepr
from claveles.circuit.ops.std import BeamSplitter, MeasureHomodyne
from claveles.circuit.state import QuantumState
from claveles.circuit.visualize import make_figure
from claveles.client import SimulatorClient
from claveles.execute import execute
from numpy import linspace
```

```python
r = 3.0  # squeezing parameter
n_shots = 2000
hbar = 1.0

xvec = linspace(-10, 10, 401)
pvec = linspace(-10, 10, 401)

sns.set_theme(style="whitegrid")
```

---

## 2. Generating the EPR State

As described above, a physical EPR state can be generated by interfering two single-mode squeezed vacuum states ($x$-squeezed for mode 0 and $p$-squeezed for mode 1) on a 50:50 beamsplitter with a phase of $\phi = \pi$. This phase configuration directly establishes the EPR coupling, eliminating the need for subsequent phase rotations.

```python
def build_epr_bs(r: float, name: str = "EPR via BeamSplitter") -> CircuitRepr:
    c = CircuitRepr(name)
    c.Q(0) | QuantumState.squeezed(r=r, phi=0)  # x-squeezed
    c.Q(1) | QuantumState.squeezed(r=r, phi=pi / 2)  # p-squeezed
    c.Q(0, 1) | BeamSplitter(theta=pi / 4, phi=pi)
    return c
```

```python
circuit_bs = build_epr_bs(r=r)
fig = make_figure(circuit_bs)
fig.show()
```

---

## 3. Visualizing Individual-Mode Wigner Functions

Prior to evaluating the joint correlations, we visualize the quantum state of each mode individually by *tracing out* (discarding) the other mode. A single-mode subsystem of an EPR state represents a thermal state; its Wigner function manifests as an isotropic Gaussian distribution that is significantly broader than the vacuum state in all phase-space directions.

```python
client = SimulatorClient(backend="cpu", state_save_policy="first_only")
client.n_shots = 1

result_bs = execute(circuit_bs, client)
final_state_bs = result_bs.client_result.shot_results[0].state

for mode_idx, label in [(0, "mode 0"), (1, "mode 1")]:
    state = final_state_bs.extract_mode(mode_idx)
    fig = state.draw_wigner(xvec, pvec, title=f"EPR (BS method) — {label} marginal")
    plt.show()
```

The marginal distribution of each mode is a broad, isotropic Gaussian, which is the characteristic signature of a thermal state. In contrast, the initial squeezed input states possessed highly anisotropic (elliptical) Wigner functions. The beamsplitter successfully converts single-mode squeezing into inter-mode quantum correlations, thereby generating entanglement. Tracing out one mode discards these correlations, explaining why neither mode appears squeezed when analyzed in isolation.

Importantly, these correlations are not destroyed; rather, tracing out one mode renders them unobservable. They remain encoded within the joint state and can only be accessed via joint measurements that probe both modes simultaneously.

## 4. Quantum Correlations via Homodyne Measurement

### 4.1 Measuring the Joint Distributions

We introduce homodyne measurement operations into the circuit and collect $n_\text{shots}$ experimental samples. Alongside the correlated quadrature pairs $(x_0, x_1)$ and $(p_0, p_1)$, we also record the cross-combination $(x_0, p_1)$ to serve as a control sample. This control pairs conjugate quadratures across different modes, where no EPR correlations are theoretically expected.

```python
def build_epr_bs_with_homodyne(r: float, measure_x: bool = True) -> CircuitRepr:
    c = build_epr_bs(r=r, name=f"EPR BS {'x' if measure_x else 'p'} corr")
    if measure_x:
        c.Q(0) | MeasureHomodyne(phi=0)  # measure x
        c.Q(1) | MeasureHomodyne(phi=0)  # measure x
    else:
        c.Q(0) | MeasureHomodyne(phi=pi / 2)  # measure p
        c.Q(1) | MeasureHomodyne(phi=pi / 2)  # measure p
    return c


def build_epr_bs_cross(r: float) -> CircuitRepr:
    c = build_epr_bs(r=r, name="EPR BS cross")
    c.Q(0) | MeasureHomodyne(phi=0)  # measure x of mode 0
    c.Q(1) | MeasureHomodyne(phi=pi / 2)  # measure p of mode 1
    return c
```

```python
client_shots = SimulatorClient(backend="cpu", state_save_policy="none")
client_shots.n_shots = n_shots

circuit_x = build_epr_bs_with_homodyne(r=r, measure_x=True)
result_x = execute(circuit_x, client_shots)
x0 = np.array([result_x.client_result[i].get_value(0) for i in range(n_shots)])
x1 = np.array([result_x.client_result[i].get_value(1) for i in range(n_shots)])

circuit_p = build_epr_bs_with_homodyne(r=r, measure_x=False)
result_p = execute(circuit_p, client_shots)
p0 = np.array([result_p.client_result[i].get_value(0) for i in range(n_shots)])
p1 = np.array([result_p.client_result[i].get_value(1) for i in range(n_shots)])

circuit_cross = build_epr_bs_cross(r=r)
result_cross = execute(circuit_cross, client_shots)
x0_cross = np.array(
    [result_cross.client_result[i].get_value(0) for i in range(n_shots)]
)
p1_cross = np.array(
    [result_cross.client_result[i].get_value(1) for i in range(n_shots)]
)
```

### 4.2 Scatter Plots of the Joint Distributions

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].scatter(x0, x1, s=3, alpha=0.3, color="steelblue")
axes[0].set_xlabel(r"$x_0$")
axes[0].set_ylabel(r"$x_1$")
axes[0].set_title(r"Joint distribution of $(x_0, x_1)$")
axes[0].set_aspect("equal")
axes[0].axline(
    (0, 0), slope=1, color="red", linestyle="--", linewidth=1, label=r"$x_0 = x_1$"
)
axes[0].legend()

axes[1].scatter(p0, p1, s=3, alpha=0.3, color="darkorange")
axes[1].set_xlabel(r"$p_0$")
axes[1].set_ylabel(r"$p_1$")
axes[1].set_title(r"Joint distribution of $(p_0, p_1)$")
axes[1].set_aspect("equal")
axes[1].axline(
    (0, 0), slope=-1, color="red", linestyle="--", linewidth=1, label=r"$p_0 = -p_1$"
)
axes[1].legend()

axes[2].scatter(x0_cross, p1_cross, s=3, alpha=0.3, color="gray")
axes[2].set_xlabel(r"$x_0$")
axes[2].set_ylabel(r"$p_1$")
axes[2].set_title(r"Control: $(x_0, p_1)$, no correlation")
axes[2].set_aspect("equal")

fig.suptitle(f"EPR correlations (BS method, $r = {r}$)")
fig.tight_layout()
plt.show()

print(f"Corr(x0, p1) = {np.corrcoef(x0_cross, p1_cross)[0, 1]:.4f}  (expected: ~0)")
```

The left panel demonstrates that the $(x_0, x_1)$ samples cluster tightly along the diagonal line $x_0 = x_1$, showing that the two $x$-quadratures are strongly correlated. The middle panel shows that the $(p_0, p_1)$ samples align along the anti-diagonal line $p_0 = -p_1$, confirming strong anti-correlation between the $p$-quadratures. These behaviors constitute the defining signatures of an EPR state. The right panel serves as our control: the cross-quadrature distribution $(x_0, p_1)$ is completely isotropic with no directional structure, and its correlation coefficient is close to zero. This confirms that EPR correlations exist selectively between $x_0$–$x_1$ and $p_0$–$p_1$, while conjugate cross-quadratures carry entirely independent information.

## 5. Conditional State After Homodyne Measurement

As introduced in the background section, performing a homodyne measurement on one mode of an EPR pair projects the remaining mode into a displaced squeezed state based on the measurement outcome. Let us verify this phenomenon directly: a measurement outcome of $x_0 = a$ should project mode 1 into a state centered near $\langle \hat{x}_1 \rangle \approx a$.

Below, we visualize the conditional state of mode 1 given a specific measurement outcome on mode 0.

```python
def build_epr_bs_conditional(r: float) -> CircuitRepr:
    c = build_epr_bs(r=r, name="EPR BS conditional")
    c.Q(0) | MeasureHomodyne(phi=0)  # measure x of mode 0
    return c
```

```python
circuit_cond = build_epr_bs_conditional(r=r)

client_cond = SimulatorClient(backend="cpu", state_save_policy="first_only")
client_cond.n_shots = 1

result_cond = execute(circuit_cond, client_cond)
meas_val = result_cond.client_result[0].get_value(0)
state_mode1 = result_cond.client_result.shot_results[0].state.extract_mode(1)

print(f"Measurement outcome on mode 0: x0 = {meas_val:.4f}")
print(f"Conditional mean of mode 1:    x1 = {state_mode1.mean[0].real:.4f}")

# Center the x grid on the outcome so that the displaced state stays in view.
fig = state_mode1.draw_wigner(
    xvec + meas_val,
    pvec,
    title=f"Conditional state of mode 1 (measured x0 = {meas_val:.3f})",
)
ax = fig.axes[0]
ax.axvline(
    meas_val,
    color="red",
    linestyle="--",
    linewidth=1.2,
    label=f"measured x0 = {meas_val:.3f}",
)
ax.legend(loc="upper right")
fig.tight_layout()
plt.show()
```

The conditional state of mode 1 is an $x$-squeezed state displaced to approximately $x_1 \approx x_0$. Repeating this experiment with varying measurement outcomes yields states with different displacements but an identical $x$-squeezed profile. The measurement process randomly selects which displaced state manifests, which is fully consistent with the underlying EPR correlation $\hat{x}_0 \approx \hat{x}_1$.

## Summary

This tutorial demonstrated the following:

1. **Generation by beamsplitter**: A two-mode EPR state is produced by mixing an $x$-squeezed vacuum and a $p$-squeezed vacuum on a 50:50 beamsplitter with phase $\phi = \pi$.
2. **Single-mode marginal is a thermal state**: Tracing out one mode leaves the remaining mode in a mixed thermal state (characterized by an isotropic Gaussian distribution), illustrating that the entanglement is stored entirely within the inter-mode correlations.
3. **EPR correlations**: The joint distributions of $(x_0, x_1)$ and $(p_0, p_1)$ reveal strong correlations ($\hat{x}_0 \approx \hat{x}_1$ and $\hat{p}_0 \approx -\hat{p}_1$), whereas the cross-quadrature combination $(x_0, p_1)$ shows no correlation, demonstrating that the quantum correlations are highly selective.
4. **Conditional state after measurement**: Measuring one mode projects the other into a displaced squeezed state, where the displacement is determined by the specific measurement outcome in accordance with the projection postulate.
